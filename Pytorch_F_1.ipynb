```python
import torch
import torch as pt
torch.__version__
```

```python
# intro to tensor
# ther are scaler vector and matrix tensors bacically 1d 2d and 3d tensors

scaler=torch.tensor(2)
print(scaler)
print(scaler.item()) # give us the no. in the tensor in form of int
print(scaler.ndim) # gives us the dimensions
print(scaler.itemsize) # returns the bytes
```

```python
# vetors
vector=torch.tensor([2,3])

# matrix
matrix=torch.tensor([[1,2,3],
                    [4,5,6]])

print(vector.shape)
print(matrix.shape)
```

```python
# Tensor
# Tensor a big martrix in which we can add mutiple matrix

TENSOR=torch.tensor([[[1,2,3],[4,5,6]], # matrix one of 2x3
                     [[2,2,2],[2,2,2]]]) # martix 2 of 2x3
print(TENSOR.shape)
print(TENSOR.ndim)
TENSOR[1]
```

```python
t=pt.tensor([[[1,2,3],[1,2,3],[1,2,4]]
             ,[[1,2,3],[1,2,4],[1,2,3]]])
print(t.shape)
print(t.ndim)
t[0]
```
